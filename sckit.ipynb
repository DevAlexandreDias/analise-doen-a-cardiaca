```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
     

models = {"KNN": KNeighborsClassifier(),
           "Logistc Regression": LogisticRegression(max_iter=1000),
           "Random Forest": RandomForestClassifier()}

def fit_and_score(models, X_train, X_test, y_train, y_test):
  np.random.seed(42)

  model_scores = {}

  for name, model in models.items():
    model.fit(X_train, y_train)
    model_scores[name] = model.score(X_test, y_test)
  return model_scores
resultados = fit_and_score(models, X_train, X_test, y_train, y_test)
print(resultados)

```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# -------------------------------------------------------------------------
# 1. PREPARAÇÃO DOS DADOS (Simulação para o exemplo)
# -------------------------------------------------------------------------
# 'y_probs' são as CHANCES (de 0% a 100%) que o modelo deu para cada paciente ter doença.
# 'y_test' é a REALIDADE: 0 significa "Saudável" e 1 significa "Com doença".
np.random.seed(42)
y_probs = np.concatenate(
    [np.random.beta(2, 5, 50), np.random.beta(5, 2, 50)]
)  # Cria probabilidades
y_test = np.array([0] * 50 + [1] * 50)  # 50 pessoas saudáveis e 50 doentes

# -------------------------------------------------------------------------
# 2. CRIAÇÃO DO GRÁFICO
# -------------------------------------------------------------------------
# Define o tamanho da imagem da figura (10 de largura por 6 de altura)
plt.figure(figsize=(10, 6))

# O 'sns.histplot' desenha as colunas (histograma) e as curvas de tendência:
# - x: As chances calculadas pelo modelo (eixo horizontal).
# - hue: Separa as cores pela realidade do paciente (0 = Verde, 1 = Vermelho).
# - kde=True: Desenha a linha suave em cima das barras para mostrar a tendência.
# - bins=20: Divide o eixo X em 20 "faixas" de probabilidade (ex: 0-5%, 5-10%...).
sns.histplot(
    x=y_probs,
    hue=y_test,
    kde=True,
    bins=20,
    palette={0: "green", 1: "red"},
    element="step",  # Deixa as colunas mais fáceis de visualizar sem cobrir uma à outra
)

# -------------------------------------------------------------------------
# 3. LINHA DE CORTE (O "Muro" de Decisão do Modelo)
# -------------------------------------------------------------------------
# Desenha uma linha preta tracejada exatamente no meio (0.5 ou 50%).
# O modelo usa essa linha como regra:
# - Se a chance for MAIOR que 0.5 (à direita) -> Ele chuta que a pessoa ESTÁ doente.
# - Se a chance for MENOR que 0.5 (à esquerda) -> Ele chuta que a pessoa está SAUDÁVEL.
plt.axvline(
    x=0.5,
    color="black",
    linestyle="--",
    linewidth=2,
    label="Fronteira de Decisão (50%)",
)

# -------------------------------------------------------------------------
# 4. TEXTOS E EXPLICAÇÕES NO GRÁFICO (Para quem está lendo entender rápido)
# -------------------------------------------------------------------------
# Título principal do gráfico
plt.title(
    "Como o Modelo Avalia a Chance de Doença Cardíaca",
    fontsize=14,
    fontweight="bold",
)

# Nome do Eixo X (Horizontal): A chance estimada
plt.xlabel(
    "Chance Prevista pelo Modelo (0.0 = 0% de chance | 1.0 = 100% de chance)",
    fontsize=11,
)

# Nome do Eixo Y (Vertical): Quantas pessoas caíram naquela faixa
plt.ylabel("Quantidade de Pacientes", fontsize=11)

# Ajusta a legenda para indicar o que cada cor e a linha significam
plt.legend(
    labels=[
        "Fronteira (50%)",
        "Pacientes Realmente Doentes (1)",
        "Pacientes Realmente Saudáveis (0)",
    ]
)

# Mostra o gráfico na tela
plt.show()
plt.show()
```
